````python
#@title Herramientas { run: "auto", display-mode: "form" }

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.transforms as mtransforms
from matplotlib.patches import Circle, Polygon, Rectangle
from cycler import cycler
from fractions import Fraction
from io import StringIO
import scipy
from scipy import stats

# La paleta de las ilustraciones del apunte, para que los gráficos se vean como el mismo material.
CREMA, TEAL, TEAL_PALIDO, MOSTAZA, CORAL, TINTA = (
    "#FEF8EF", "#61A9A1", "#D0E5DE", "#EEA92F", "#F07C52", "#32435D")

plt.rcParams.update({
    "figure.facecolor": CREMA, "axes.facecolor": CREMA, "savefig.facecolor": CREMA,
    "axes.edgecolor": TINTA, "axes.labelcolor": TINTA, "text.color": TINTA,
    "xtick.color": TINTA, "ytick.color": TINTA,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.prop_cycle": cycler(color=[TEAL, CORAL, MOSTAZA, TINTA]),
    "axes.grid": True, "grid.color": TEAL_PALIDO, "grid.linewidth": 0.8, "axes.axisbelow": True,
    "figure.figsize": (9, 4.5), "figure.dpi": 100,
    "axes.titlesize": 13, "axes.titleweight": "bold", "axes.titlelocation": "left",
    "legend.frameon": False, "font.size": 11,
})


def num(x, decimales=4):
    """Número con coma decimal, como en el apunte."""
    if x is None or (isinstance(x, float) and np.isnan(x)):
        return "—"
    if abs(x) < 0.5 * 10 ** -decimales:
        x = 0.0   # evita el "-0,0000"
    return f"{x:,.{decimales}f}".replace(",", "X").replace(".", ",").replace("X", ".")


def frac(k, n):
    """'k/n = 0,xxxx' con la fracción sin simplificar, que es como la escribe el apunte."""
    return f"{k}/{n} = {num(k / n)}"


print("numpy", np.__version__, "| pandas", pd.__version__, "| scipy", scipy.__version__, "| todo listo")


from pathlib import Path

_num_de_un_valor = num


def num(x, decimales=4):
    """Número con coma decimal y punto de miles, como en el apunte. Con una columna, lo hace en cada valor."""
    if isinstance(x, pd.Series):
        return x.map(lambda valor: _num_de_un_valor(valor, decimales))
    return _num_de_un_valor(x, decimales)


def formato_eje(ax):
    """Escribe el eje horizontal con coma decimal y punto de miles, y con pocas marcas para que no se pisen."""
    ax.xaxis.set_major_locator(plt.MaxNLocator(5))
    decimales = 0 if ax.get_xlim()[1] >= 10000 else 1
    ax.xaxis.set_major_formatter(lambda valor, _: num(valor, decimales))


def _valores(columna):
    return pd.Series(columna).dropna().to_numpy(dtype=float)


# Con la misma semilla, una tanda corta es el comienzo de una más larga: se guarda la más larga.
_SORTEOS = {}


def _sortear(cantidad, n, repeticiones, semilla):
    """Las posiciones de muchas muestras aleatorias simples de tamaño n, sin reposición: una fila por muestra."""
    clave = (cantidad, n, semilla)
    if len(_SORTEOS.get(clave, [])) < repeticiones:
        rs = np.random.RandomState(semilla)
        posiciones = np.empty((repeticiones, n), dtype=np.int32)
        for fila in range(repeticiones):
            posiciones[fila] = rs.choice(cantidad, n, replace=False)
        _SORTEOS[clave] = posiciones
    return _SORTEOS[clave][:repeticiones]


def medias_de_muestras(columna, n, repeticiones=2000, semilla=8102):
    """La media de cada una de muchas muestras aleatorias simples de tamaño n."""
    x = _valores(columna)
    return pd.Series(x[_sortear(len(x), n, repeticiones, semilla)].mean(axis=1))


def _histograma_de_medias(ax, medias, mu, titulo):
    ax.hist(medias, bins=50, color=TEAL, edgecolor=CREMA)
    ax.axvline(mu, color=TINTA, linewidth=1.6)
    ax.set_title(titulo)
    ax.set_ylabel("cantidad de muestras")


def dibujar_medias(medias, mu, titulo):
    """El histograma de muchas medias muestrales, con una línea en el valor verdadero."""
    fig, ax = plt.subplots()
    _histograma_de_medias(ax, medias, mu, titulo)
    ax.set_xlabel("media de cada muestra; la línea es μ")
    formato_eje(ax)
    plt.show()


def dibujar_dos_medias(medias_1, medias_2, mu, titulo_1, titulo_2):
    """Dos histogramas de medias, uno arriba del otro y con el mismo eje, para comparar el ancho."""
    fig, (arriba, abajo) = plt.subplots(2, 1, figsize=(9, 7), sharex=True)
    _histograma_de_medias(arriba, medias_1, mu, titulo_1)
    _histograma_de_medias(abajo, medias_2, mu, titulo_2)
    abajo.set_xlabel("media de cada muestra; la línea es μ")
    formato_eje(abajo)
    fig.tight_layout()
    plt.show()


def varianzas_de_muestras(columna, n, ddof, repeticiones=10000, semilla=8102):
    """La varianza de cada una de muchas muestras de tamaño n, con divisor n - ddof."""
    x = _valores(columna)
    return pd.Series(x[_sortear(len(x), n, repeticiones, semilla)].var(axis=1, ddof=ddof))


def intervalos_de_muestras(columna, n, confianza=0.95, repeticiones=100, semilla=8102):
    """El intervalo de confianza para la media, con la t, de cada una de muchas muestras."""
    x = _valores(columna)
    muestras = x[_sortear(len(x), n, repeticiones, semilla)]
    m = muestras.mean(axis=1)
    e = stats.t.ppf(0.5 + confianza / 2, n - 1) * muestras.std(axis=1, ddof=1) / np.sqrt(n)
    return pd.DataFrame({"media": m, "inferior": m - e, "superior": m + e})


def dibujar_intervalos(intervalos, mu, titulo):
    """Un segmento por intervalo: verde azulado si cubre a mu, coral si no la cubre."""
    cubre = ((intervalos["inferior"] <= mu) & (mu <= intervalos["superior"])).to_numpy()
    fig, ax = plt.subplots(figsize=(10, 7))
    for k, (m, a, b) in enumerate(intervalos[["media", "inferior", "superior"]].to_numpy()):
        color = TEAL if cubre[k] else CORAL
        ax.plot([a, b], [k, k], color=color, linewidth=1.3 if cubre[k] else 2.4)
        ax.plot(m, k, "o", color=color, markersize=2.5)
    ax.axvline(mu, color=TINTA, linewidth=1.6)
    ax.set_yticks([])
    ax.grid(axis="y", visible=False)
    ax.set_xlabel("un segmento por muestra: su intervalo del 95%; la línea es μ")
    formato_eje(ax)
    ax.set_title(f"{titulo}: {cubre.sum()} de {len(cubre)} cubren el valor verdadero")
    plt.show()


def intervalo_t(columna, confianza=0.95):
    """El intervalo de confianza para la media de una muestra, con la t: (inferior, superior)."""
    x = _valores(columna)
    e = stats.t.ppf(0.5 + confianza / 2, len(x) - 1) * x.std(ddof=1) / np.sqrt(len(x))
    return x.mean() - e, x.mean() + e


def tabla_de_intervalos(tabla, columna, tamanios, mu, semilla=8102):
    """Una muestra de cada tamaño, sacada de la tabla, con su intervalo del 95% para la media."""
    if isinstance(tabla, str):
        raise TypeError(f'Escribiste "{tabla}" entre comillas. Ahí va el nombre de la tabla, sin comillas.')
    if len(tabla) < max(tamanios):
        raise ValueError(f"La tabla tiene {len(tabla)} filas y no alcanza para una muestra de "
                         f"{max(tamanios)}. Revisá el texto de la condición: mayúsculas, tildes y espacios.")
    filas = []
    for n in tamanios:
        x = tabla.sample(n=n, random_state=semilla)[columna]
        a, b = intervalo_t(x)
        filas.append({"n": num(n, 0), "media de la muestra": num(x.mean(), 0),
                      "intervalo del 95%": f"[{num(a, 0)} ; {num(b, 0)}]",
                      "ancho": num(b - a, 0), "¿cubre a μ?": "sí" if a <= mu <= b else "no"})
    return pd.DataFrame(filas).set_index("n")


def proporcion_de_rechazos(columna, mu0, n, alfa=0.05, repeticiones=1000, semilla=8102):
    """En qué proporción de muchas muestras el test de la media rechaza H₀: μ = mu0."""
    x = _valores(columna)
    p = stats.ttest_1samp(x[_sortear(len(x), n, repeticiones, semilla)], mu0, axis=1).pvalue
    return float(np.mean(p <= alfa))


def comparar_disenos(tabla, columna, n=200, repeticiones=10000, semilla=8102):
    """Saca muchas muestras con cuatro diseños y compara cuánto varían sus estimaciones de la media."""
    datos = tabla.dropna(subset=[columna])
    x = datos[columna].to_numpy(dtype=float)
    rs = np.random.RandomState(semilla)

    def estratificado(variable):
        grupos = [g.to_numpy(dtype=float) for _, g in datos.groupby(variable)[columna]]
        pesos = [len(g) / len(x) for g in grupos]
        tamanios = [max(2, round(n * w)) for w in pesos]
        adentro = sum(w * g.var() for g, w in zip(grupos, pesos)) / x.var()

        def una():
            m = sum(w * g[rs.choice(len(g), k, replace=False)].mean()
                    for g, w, k in zip(grupos, pesos, tamanios))
            return m, sum(tamanios)
        return una, adentro

    hogares = [g.to_numpy(dtype=float) for _, g in datos.groupby("id_hogar")[columna]]
    cantidad = round(n * len(hogares) / len(x))

    def conglomerados():
        v = np.concatenate([hogares[i] for i in rs.choice(len(hogares), cantidad, replace=False)])
        return v.mean(), len(v)

    disenos = {
        "Aleatorio simple": (lambda: (x[rs.choice(len(x), n, replace=False)].mean(), n), None),
        "Estratificado por región": estratificado("region"),
        "Estratificado por nivel educativo": estratificado("nivel_educativo"),
        f"Por conglomerados: {cantidad} hogares enteros": (conglomerados, None),
    }
    filas, base = [], None
    for nombre, (sacar, adentro) in disenos.items():
        resultado = np.array([sacar() for _ in range(repeticiones)])
        varianza = resultado[:, 0].var()
        base = base or varianza
        filas.append({"diseño": nombre, "casos por muestra": num(resultado[:, 1].mean(), 0),
                      "desvío de las estimaciones": num(np.sqrt(varianza), 0),
                      "efecto de diseño": num(varianza / base, 2),
                      "varianza que queda adentro de los estratos":
                          "—" if adentro is None else num(100 * adentro, 1) + "%"})
    return pd.DataFrame(filas).set_index("diseño")
````

````python
#@title Controles de los ejercicios { run: "auto", display-mode: "form" }

import re

# Las respuestas se calculan acá, a partir de los datos, y no se muestran.
# Cada prueba es un texto y una función sin argumentos, para que un error en una no corte las demás.
# Una prueba devuelve si la línea está bien, o un texto que dice qué revisar.
# Cada línea se revisa con los valores que el estudiante calculó antes: un error no arrastra a las que siguen.
_POB = ocupados[ocupados["ingreso"] != -9]


def _cerca(a, b, tolerancia=1e-6):
    """Iguales salvo redondeo. Un monto de mil o más admite medio peso, para quien redondea a pesos."""
    if isinstance(a, str):
        return False
    try:
        a, b = float(a), float(b)
    except (TypeError, ValueError):
        return False
    if a == b:
        return True
    margen = tolerancia * max(1.0, abs(b))
    if abs(b) >= 1000:
        margen = max(margen, 0.5)
    return abs(a - b) <= margen


def _misma_tabla(a, b):
    try:
        return len(a) == len(b) and a.index.equals(b.index)
    except Exception:
        return False


def _ingreso_de(g, sexo):
    m = g["muestra"]
    return m.loc[m["sexo"] == sexo, "ingreso"]


def _pruebas_0(g):
    noreste = ocupados[ocupados["region"] == "Noreste"]
    yield "`cantidad_noreste`: las filas de `noreste`", lambda: g["cantidad_noreste"] == len(noreste)
    yield "`edad_media_noreste`", lambda: _cerca(g["edad_media_noreste"], noreste["edad"].mean())


def _pruebas_1(g):
    yield "`cantidad_sin_respuesta`: cuántos -9 hay", lambda: g["cantidad_sin_respuesta"] == (
        ocupados["ingreso"] == -9).sum()
    yield "`poblacion` no tiene el -9 y conserva los ceros", lambda: _misma_tabla(g["poblacion"], _POB)
    yield "`N`: las filas de `poblacion`", lambda: g["N"] == len(_POB)
    yield "`mu`", lambda: _cerca(g["mu"], _POB["ingreso"].mean())
    yield "`sigma`, con divisor N porque es la población", lambda: _cerca(g["sigma"], _POB["ingreso"].std(ddof=0))
    yield "`p_mujeres`", lambda: _cerca(g["p_mujeres"], (_POB["sexo"] == "mujer").mean())


def _pruebas_2(g):
    yield "`muestra`: 200 ocupados sorteados con la semilla 8102", lambda: _misma_tabla(
        g["muestra"], g["poblacion"].sample(n=200, random_state=8102))
    yield "`n`", lambda: g["n"] == 200
    yield "`x_barra`", lambda: _cerca(g["x_barra"], g["muestra"]["ingreso"].mean())
    yield "`s`, con divisor n − 1 porque es una muestra", lambda: _cerca(g["s"], g["muestra"]["ingreso"].std())
    yield "`p_sombrero`", lambda: _cerca(g["p_sombrero"], (g["muestra"]["sexo"] == "mujer").mean())


def _pruebas_3(g):
    yield "`medias_25`: 2.000 medias de muestras de 25 ingresos", lambda: _cerca(
        g["medias_25"].mean(), medias_de_muestras(g["poblacion"]["ingreso"], n=25).mean())
    yield "`medias_100`: 2.000 medias de muestras de 100 ingresos", lambda: _cerca(
        g["medias_100"].mean(), medias_de_muestras(g["poblacion"]["ingreso"], n=100).mean())
    yield "`ee_25`: `sigma` sobre la raíz de 25", lambda: _cerca(g["ee_25"], g["sigma"] / 5)
    yield "`ee_100`: `sigma` sobre la raíz de 100", lambda: _cerca(g["ee_100"], g["sigma"] / 10)


def _pruebas_4(g):
    horas = lambda: g["poblacion"]["horas_semanales"]
    yield "`sigma2_horas`: la varianza de la población", lambda: _cerca(g["sigma2_horas"], horas().var(ddof=0))
    yield "`varianzas_n`: divisor n", lambda: _cerca(
        g["varianzas_n"].mean(), varianzas_de_muestras(horas(), n=10, ddof=0).mean())
    yield "`varianzas_n_menos_1`: divisor n − 1", lambda: _cerca(
        g["varianzas_n_menos_1"].mean(), varianzas_de_muestras(horas(), n=10, ddof=1).mean())


def _pruebas_5(g):
    def pesos_bien():
        conteo = g["poblacion"]["region"].value_counts()
        return all(_cerca(g["pesos"].get(k), v) for k, v in (conteo / conteo.sum()).items())

    def medias_bien():
        medias = g["muestra"].groupby("region")["ingreso"].mean()
        return all(_cerca(g["medias_region"].get(k), v) for k, v in medias.items())
    yield "`pesos`: el peso de cada región en la población", pesos_bien
    yield "`medias_region`: la media de cada región en la muestra", medias_bien
    yield "`x_estratos`: cada peso por la media de su región", lambda: _cerca(
        g["x_estratos"], (g["pesos"] * g["medias_region"]).sum())


def _pruebas_6(g):
    yield "`ee`: el error standard, `s` sobre la raíz de `n`", lambda: _cerca(g["ee"], g["s"] / np.sqrt(g["n"]))
    yield "`t_critico`: deja un 2,5% en cada cola", lambda: abs(
        g["t_critico"] - stats.t.ppf(0.975, g["n"] - 1)) <= 5e-5
    yield "`e`: el margen de error", lambda: _cerca(g["e"], g["t_critico"] * g["ee"])
    yield "`inferior` y `superior`", lambda: _cerca(g["inferior"], g["x_barra"] - g["e"]) and _cerca(
        g["superior"], g["x_barra"] + g["e"])


def _pruebas_7(g):
    def cobertura(n, repeticiones):
        tabla = intervalos_de_muestras(g["poblacion"]["ingreso"], n=n, repeticiones=repeticiones)
        return ((tabla["inferior"] <= g["mu"]) & (g["mu"] <= tabla["superior"])).mean()
    yield "`cubre_100`: cuántos de los 100 intervalos cubren a μ", lambda: _cerca(
        g["cubre_100"].sum(), 100 * cobertura(200, 100))
    yield "`cobertura_200`: 5.000 intervalos con n = 200", lambda: _cerca(g["cobertura_200"], cobertura(200, 5000))
    yield "`cobertura_30`: 5.000 intervalos con n = 30", lambda: _cerca(g["cobertura_30"], cobertura(30, 5000))


def _pruebas_8(g):
    def intervalo_bien():
        # Se acepta el z de la tabla, 1,96, además del que da stats.norm.ppf(0.975).
        return any(_cerca(g["inferior_p"], g["p_sombrero"] - z * g["ee_p"])
                   and _cerca(g["superior_p"], g["p_sombrero"] + z * g["ee_p"])
                   for z in (stats.norm.ppf(0.975), 1.96))
    yield "`ee_p`: el error standard con p̂", lambda: _cerca(
        g["ee_p"], np.sqrt(g["p_sombrero"] * (1 - g["p_sombrero"]) / g["n"]))
    yield "`inferior_p` y `superior_p`", intervalo_bien
    yield "`p0`: la proporción que afirma H₀", lambda: _cerca(g["p0"], 0.5)
    yield "`ee_p0`: el error standard con p₀", lambda: _cerca(g["ee_p0"], np.sqrt(g["p0"] * (1 - g["p0"]) / g["n"]))
    yield "`z_obs`", lambda: _cerca(g["z_obs"], (g["p_sombrero"] - g["p0"]) / g["ee_p0"])
    yield "`valor_p`", lambda: _cerca(g["valor_p"], 2 * stats.norm.sf(abs(g["z_obs"])))


def _pruebas_9(g):
    yield "`n_100mil`: con `sigma` y un margen de 100000", lambda: _cerca(
        g["n_100mil"], np.ceil((1.96 * g["sigma"] / 100000) ** 2))
    yield "`n_50mil`: con un margen de 50000", lambda: _cerca(g["n_50mil"], np.ceil((1.96 * g["sigma"] / 50000) ** 2))
    yield "`n_50mil_corregido`: con el tamaño `N` de la población", lambda: _cerca(
        g["n_50mil_corregido"], np.ceil(g["n_50mil"] / (1 + (g["n_50mil"] - 1) / g["N"])))


def _pruebas_10(g):
    def marco():
        p = g["poblacion"]
        return p[p["categoria_ocupacional"] == "obrero o empleado"]
    yield "`marco_asalariados`: los ocupados en relación de dependencia", lambda: _misma_tabla(
        g["marco_asalariados"], marco())
    yield "`tabla_sesgo`: muestras de 100, 1.000 y 5.000 del marco de asalariados", lambda: g["tabla_sesgo"].equals(
        tabla_de_intervalos(marco(), "ingreso", [100, 1000, 5000], g["mu"]))
    yield "`media_del_marco`", lambda: _cerca(g["media_del_marco"], marco()["ingreso"].mean())


def _pruebas_11(g):
    yield "`mu0`: el valor que afirma el informe", lambda: _cerca(g["mu0"], 650000)
    yield "`t_obs`: dividido por el error standard `ee`", lambda: _cerca(
        g["t_obs"], (g["x_barra"] - g["mu0"]) / g["ee"])
    yield "`valor_p_media`: con n − 1 grados de libertad", lambda: _cerca(
        g["valor_p_media"], 2 * stats.t.sf(abs(g["t_obs"]), g["n"] - 1))
    yield "`alfa`: el nivel de significación del enunciado", lambda: _cerca(g["alfa"], 0.05)
    yield "`prueba`: el mismo test con scipy", lambda: _cerca(
        g["prueba"].pvalue, stats.ttest_1samp(g["muestra"]["ingreso"], g["mu0"]).pvalue)


def _pruebas_12(g):
    x = lambda: g["poblacion"]["ingreso"]
    yield "`tipo_1`: con μ₀ igual a la media verdadera, `mu`", lambda: _cerca(
        g["tipo_1"], proporcion_de_rechazos(x(), mu0=g["mu"], n=200))
    yield "`potencia_200`", lambda: _cerca(g["potencia_200"], proporcion_de_rechazos(x(), mu0=650000, n=200))
    yield "`potencia_50`", lambda: _cerca(g["potencia_50"], proporcion_de_rechazos(x(), mu0=650000, n=50))
    yield "`potencia_800`", lambda: _cerca(g["potencia_800"], proporcion_de_rechazos(x(), mu0=650000, n=800))


def _pruebas_13(g):
    yield "`varones` y `mujeres`", lambda: _misma_tabla(g["varones"], _ingreso_de(g, "varón")) and _misma_tabla(
        g["mujeres"], _ingreso_de(g, "mujer"))
    yield "`welch`: el test de Welch, sin suponer varianzas iguales", lambda: _cerca(
        g["welch"].pvalue, stats.ttest_ind(g["varones"], g["mujeres"], equal_var=False).pvalue)
    yield "`dif`", lambda: _cerca(g["dif"], g["varones"].mean() - g["mujeres"].mean())
    yield "`ee_dif`", lambda: _cerca(
        g["ee_dif"], np.sqrt(g["varones"].var() / len(g["varones"]) + g["mujeres"].var() / len(g["mujeres"])))
    yield "`inferior_dif` y `superior_dif`", lambda: all(
        _cerca(g[nombre], g["dif"] + signo * stats.t.ppf(0.975, g["welch"].df) * g["ee_dif"])
        for nombre, signo in (("inferior_dif", -1), ("superior_dif", 1)))


def _pruebas_14(g):
    yield "`pares`: 200 personas del panel, con la semilla 8102", lambda: _misma_tabla(
        g["pares"], panel.sample(n=200, random_state=8102))
    yield "`apareado`: el segundo trimestre contra el primero", lambda: _cerca(
        g["apareado"].pvalue, stats.ttest_rel(g["pares"]["ingreso_2t"], g["pares"]["ingreso_1t"]).pvalue)
    yield "`independiente`", lambda: _cerca(g["independiente"].pvalue, stats.ttest_ind(
        g["pares"]["ingreso_2t"], g["pares"]["ingreso_1t"], equal_var=False).pvalue)
    yield "`dif_media_pares`", lambda: _cerca(
        g["dif_media_pares"], (g["pares"]["ingreso_2t"] - g["pares"]["ingreso_1t"]).mean())


def _pruebas_15(g):
    def intervalo_bien(inferior, superior, columna):
        x = g["muestra"][columna].dropna()
        k = len(x) - 1
        if g[inferior] > g[superior]:
            return "quedó al revés: el extremo inferior lleva el cuantil 0.975, el más grande"
        return (_cerca(g[inferior], np.sqrt(k * x.var() / stats.chi2.ppf(0.975, k)))
                and _cerca(g[superior], np.sqrt(k * x.var() / stats.chi2.ppf(0.025, k))))
    yield "`inferior_sigma_horas` y `superior_sigma_horas`", lambda: intervalo_bien(
        "inferior_sigma_horas", "superior_sigma_horas", "horas_semanales")
    yield "`inferior_sigma_ingreso` y `superior_sigma_ingreso`", lambda: intervalo_bien(
        "inferior_sigma_ingreso", "superior_sigma_ingreso", "ingreso")
    yield "`F_muestra`: la varianza de `varones` sobre la de `mujeres`", lambda: _cerca(
        g["F_muestra"], g["varones"].var() / g["mujeres"].var())


# Qué variables de ejercicios anteriores usa cada uno, para avisar de dónde viene un error.
_USA = {2: {1: ("poblacion", "mu", "sigma", "p_mujeres")}, 3: {1: ("poblacion", "mu", "sigma")},
        4: {1: ("poblacion",)}, 5: {1: ("poblacion", "mu"), 2: ("muestra", "n", "x_barra")},
        6: {1: ("mu",), 2: ("n", "x_barra", "s")}, 7: {1: ("poblacion", "mu")},
        8: {1: ("p_mujeres",), 2: ("n", "p_sombrero")}, 9: {1: ("N", "sigma")}, 10: {1: ("poblacion", "mu")},
        11: {1: ("poblacion",), 2: ("muestra", "n", "x_barra"), 6: ("ee",)}, 12: {1: ("poblacion", "mu")},
        13: {1: ("poblacion",), 2: ("muestra",)},
        15: {1: ("poblacion", "sigma"), 2: ("muestra", "n"), 13: ("varones", "mujeres")}}
# Si una de estas falla, lo que sigue en el control sale de ella y no vale la pena mostrarlo.
_BASES = ("poblacion", "muestra", "marco_asalariados", "varones", "pares")
_HUECO = "COMPLE" + "TAR"
_ULTIMA = {"error": None, "celda": None, "faltan": {}}


def _pruebas(numero):
    return globals()[f"_pruebas_{numero}"](globals())


def _nombres(numero):
    """Las variables que crea el ejercicio: las que nombra cada prueba antes de su explicación."""
    return {nombre for texto, _ in _pruebas(numero) for nombre in re.findall(r"`(\w+)`", re.split(r"[:,]", texto)[0])}


def _origen():
    """Qué ejercicio crea cada variable de los huecos."""
    return {nombre: numero for numero in range(15, -1, -1) for nombre in _nombres(numero)}


def _asignadas(codigo):
    return set(re.findall(r"^\s*([A-Za-z_]\w*)\s*=(?!=)", codigo, re.M))


def _sin_textos(codigo):
    """El código sin comentarios ni textos entre comillas. De un f-string quedan las cuentas entre llaves."""
    def sacar(m):
        texto = m.group(2) if m.group(2) is not None else m.group(3)
        return " ".join(re.findall(r"\{([^}]*)\}", texto)) if m.group(1) else '""'
    return re.sub(r"#[^\n]*", "", re.sub(r"""(f?)(?:"([^"\n]*)"|'([^'\n]*)')""", sacar, codigo))


def _faltantes(codigo):
    """Las variables de otros ejercicios que usa la celda y que todavía no existen, con el ejercicio que las crea."""
    limpio = _sin_textos(codigo)
    usadas = set(re.findall(r"(?<![\w.])([A-Za-z_]\w*)(?!\w)(?!\s*=(?!=))", limpio)) - _asignadas(limpio)
    origen = _origen()
    return {nombre: origen[nombre] for nombre in sorted(usadas, key=lambda v: (origen.get(v, 0), v))
            if nombre in origen and nombre not in _ip.user_ns}


def _lista(cosas):
    return cosas[0] if len(cosas) == 1 else ", ".join(cosas[:-1]) + " y " + cosas[-1]


def _aviso_de_faltantes(faltan):
    nombres = _lista([f"`{nombre}`" for nombre in faltan])
    ejercicios = sorted(set(faltan.values()))
    uno = len(faltan) == 1
    return (f"Esta celda no se ejecutó porque usa {nombres}, que todavía no {'existe' if uno else 'existen'}: "
            f"{'se arma' if uno else 'se arman'} en {'el ejercicio' if len(ejercicios) == 1 else 'los ejercicios'} "
            f"{_lista([str(e) for e in ejercicios])}. Hacé primero {'ese ejercicio' if len(ejercicios) == 1 else 'esos ejercicios'}, "
            "y después ejecutá esta celda de nuevo.")


def _registrar(resultado=None):
    if resultado is not None and not re.match(r"#\s*@title Control del", resultado.info.raw_cell.lstrip()):
        celda = resultado.info.raw_cell
        _ULTIMA["error"] = resultado.error_in_exec or resultado.error_before_exec
        _ULTIMA["faltan"] = {}
        if _ULTIMA["error"] is None and re.search(rf"\b{_HUECO}\b", celda):
            _ULTIMA["error"] = "huecos"
        elif _ULTIMA["error"] is None and not celda.lstrip().startswith("#@title") and _faltantes(celda):
            _ULTIMA["error"], _ULTIMA["faltan"] = "faltan", _faltantes(celda)
        _ULTIMA["celda"] = celda


_ip = get_ipython()
_ip.events.callbacks["post_run_cell"] = [f for f in _ip.events.callbacks["post_run_cell"]
                                         if getattr(f, "__name__", "") != "_registrar"]
_ip.events.register("post_run_cell", _registrar)


def _antes_de_ejecutar(lineas):
    """Una celda con huecos, o que usa algo de un ejercicio que no se hizo, no se ejecuta: dice por qué, y
    «Ejecutar todas» sigue."""
    codigo = "".join(lineas)
    if codigo.lstrip().startswith("#@title"):
        return lineas
    cantidad = len(re.findall(rf"\b{_HUECO}\b", codigo))
    if cantidad:
        huecos, reemplazar = ("un hueco", "reemplazalo") if cantidad == 1 else (f"{cantidad} huecos", "reemplazalos")
        return [f'print("Esta celda todavía tiene {huecos} {_HUECO}, así que no se ejecutó. '
                f'Si querés hacer el ejercicio, {reemplazar} y ejecutala de nuevo.")\n']
    faltan = _faltantes(codigo)
    if faltan:
        return [f"print({_aviso_de_faltantes(faltan)!r})\n"]
    return lineas


_ip.input_transformers_post = [f for f in _ip.input_transformers_post
                               if getattr(f, "__name__", "") not in ("_avisar_huecos", "_antes_de_ejecutar")]
_ip.input_transformers_post.append(_antes_de_ejecutar)


def _evaluar(prueba):
    """(bien, nota) de una prueba. Una prueba puede devolver un texto que dice qué revisar."""
    resultado = prueba()
    if isinstance(resultado, str):
        return False, ": " + resultado
    return bool(resultado), ""


def _pistas(codigo):
    """Lo que no da error en rojo: miles con punto o con coma, decimales con coma y variables del ejemplo."""
    codigo = _sin_textos(codigo)
    miles = r"(?<![\w.])(?:[1-9]\d{1,2}(?:\.\d{3})+|[1-9](?:\.\d{3}){2,}|[1-9]\.000)(?![\d.])"
    for numero in dict.fromkeys(re.findall(miles, codigo)):
        print(f"Ojo: en la celda escribiste {numero}. En el código los miles no llevan punto: "
              f"va {numero.replace('.', '')}.\n")
    for numero in dict.fromkeys(re.findall(r"(?<![\w.,])[1-9]\d{0,2}(?:,\d{3})+(?!\d)(?!,\d)", codigo)):
        print(f"Ojo: en la celda escribiste {numero}. En el código los miles no llevan coma: "
              f"va {numero.replace(',', '')}.\n")
    for numero in dict.fromkeys(re.findall(r"(?<![\w.])0,\d+", codigo)):
        print(f"Ojo: en la celda escribiste {numero}. En el código los decimales van con punto: "
              f"va {numero.replace(',', '.')}.\n")
    origen = _origen()
    for nombre in dict.fromkeys(re.findall(r"\b[A-Za-z]\w*_(?:relev|edad)\w*\b", codigo)):
        if nombre not in _asignadas(codigo):
            sin_terminacion = re.sub(r"_(?:relev|edad)", "", nombre)
            cual = (f"la del ejercicio, casi siempre con el mismo nombre sin la terminación: `{sin_terminacion}`"
                    if sin_terminacion in origen else "la variable que nombra el texto del ejercicio")
            print(f"Ojo: en la celda usaste `{nombre}`, que es una variable del ejemplo. En los huecos va {cual}.\n")


def _dependencias_bien(previo, nombres):
    for texto, prueba in _pruebas(previo):
        if re.match(r"`(\w+)`", texto).group(1) in nombres:
            try:
                if not _evaluar(prueba)[0]:
                    return False
            except Exception:
                return False
    return True


def controlar(numero):
    g = globals()
    propios = _nombres(numero)
    origen = {nombre: previo for previo in range(numero) for nombre in _nombres(previo)}
    lineas, bien, faltan, falta = [], 0, 0, None
    for texto, prueba in _pruebas(numero):
        try:
            ok, nota = _evaluar(prueba)
        except KeyError as e:
            ok, nombre = False, e.args[0]
            if nombre in propios:
                nota, faltan, falta = f": todavía no existe la variable `{nombre}`", faltan + 1, falta or nombre
            else:
                nota = f": usa `{nombre}`, que se crea en el ejercicio {origen.get(nombre, 'anterior')} y todavía no existe"
        except Exception:
            ok, nota = False, ": quedó un resultado de otro tipo, revisá esa línea"
        if not ok and any(isinstance(g.get(nombre), str) for nombre in re.findall(r"`(\w+)`", texto)):
            nota = ": quedó un texto entre comillas; un número o una variable van sin comillas"
        lineas.append(("✔ " if ok else "✘ ") + texto + nota)
        bien += ok
        base = next((b for b in _BASES if texto.startswith(f"`{b}`")), None)
        if base and not ok:
            lineas.append(f"Lo que sigue sale de `{base}`: corregí primero esa línea.")
            break
    revisadas = len([l for l in lineas if l[:1] in "✔✘"])

    celda = _ULTIMA["celda"]
    propia = celda is not None and bool(_asignadas(celda) & propios)
    error = _ULTIMA["error"] if propia else None
    if error == "faltan":
        print(_aviso_de_faltantes(_ULTIMA["faltan"]).replace("Esta celda", "La celda de este ejercicio", 1))
        return
    ninguna = not any(nombre in g for nombre in propios)
    if error in (None, "huecos") and (faltan == revisadas or ninguna):
        if ninguna:
            falta = re.match(r"`(\w+)`", next(_pruebas(numero))[0]).group(1)
        print(f"Todavía no existe la variable `{falta}`. Completá los huecos y ejecutá la celda del ejercicio {numero}.")
        return
    if propia and (bien < revisadas or error is not None):
        _pistas(celda)
    if error == "huecos":
        print("Ojo: la celda de este ejercicio todavía tiene huecos sin completar, así que no se ejecutó. "
              "Lo que sigue viene de una ejecución anterior.\n")
    elif error is not None:
        print(f"Ojo: la celda de este ejercicio terminó con un error ({type(error).__name__}). Corregila y "
              "ejecutala de nuevo, porque lo que sigue puede venir de una ejecución anterior.\n")
    pendientes = [previo for previo, nombres in _USA.get(numero, {}).items() if not _dependencias_bien(previo, nombres)]
    for previo in pendientes:
        nombres = " y ".join(f"`{nombre}`" for nombre in _USA[numero][previo])
        print(f"Ojo: este ejercicio usa {nombres} del ejercicio {previo}, y ahí todavía hay algo sin hacer o marcado "
              f"con ✘. Hacé o corregí primero el ejercicio {previo}, y después volvé a ejecutar la celda de este "
              "ejercicio antes del control.\n")
    print("\n".join(lineas))
    print()
    if error == "huecos":
        cierre = " Completá primero los huecos."
    elif error is not None:
        cierre = " Corregí primero el error en rojo."
    elif bien < revisadas:
        cierre = " Revisá lo marcado con ✘."
    elif pendientes:
        cierre = " Las cuentas de este ejercicio están bien, pero usan un resultado que todavía no está bien."
    else:
        cierre = " Todo bien."
    print(f"Ejercicio {numero}: {bien} de {revisadas}.{cierre}")


print("Controles listos.")
````

| un resultado absurdo, sin error en rojo | quedó un número con coma o con punto de miles, o una variable del ejemplo: ver abajo |

### Cómo se llena un hueco

Se borra solo la palabra `COMPLETAR`, y lo que la rodea queda igual:

- `"COMPLETAR"`, entre comillas: va un texto, entre esas mismas comillas. `ocupados["COMPLETAR"]`
  pasa a `ocupados["edad"]`.
- `.COMPLETAR()`: va solo el nombre de la medida, porque el punto y los paréntesis ya están.
  `columna.COMPLETAR()` pasa a `columna.mean()`.
- `COMPLETAR` sin comillas: va un número, el nombre de una variable, una cuenta corta, o `True`
  (sí) o `False` (no), con mayúscula. No lleva comillas.

En el código, los decimales van con punto y los miles no llevan separador: `0.975` y `650000`, no
`0,975` ni `650.000`. Python lee la coma como una separación entre dos números. Si un resultado da
`0`, `nan`, `inf` o un número absurdo, revisá las comas, los puntos de miles, las mayúsculas y las
tildes.

Casi todas las variables de los ejemplos terminan en `_relev` o `_edad`. En los huecos no van las
del ejemplo: va la del ejercicio, casi siempre con el mismo nombre sin la terminación, o la que
nombra el texto del ejercicio. Una variable del ejemplo en un hueco no da error en rojo: da un
número equivocado, y el control lo avisa.

# 3. La distribución de la media muestral

Antes de sortear, `x̄` es una variable aleatoria: depende de qué casos salgan. Su distribución se

Con la población a mano eso se puede ver: se sacan miles de muestras y se mira cómo se reparten sus

- `np.sqrt(...)` es la raíz cuadrada. `sigma_edad / np.sqrt(25)` es el error standard que predice la

- `stats.skew(...)` es la asimetría `α₃` de la unidad 2.
- `dibujar_medias(medias, mu, título)` dibuja el histograma de las medias, con una línea en `μ`.

Con estratos, la media no se calcula promediando todos los casos juntos: se calcula la media de cada

- Al multiplicar `pesos_relev * medias_relev`, cada peso se multiplica por la media de su misma

````python
marco_relev = pd.Series({"Concepción": 900, "Córdoba": 1300, "Ramallo": 1800})
pesos_relev = marco_relev / marco_relev.sum()
medias_relev = relevamiento.groupby("localidad")["ingreso"].mean()
x_estratos_relev = (pesos_relev * medias_relev).sum()

display(pd.DataFrame({"peso W_h": num(pesos_relev, 4), "media x̄_h": num(medias_relev, 0)}))
print("Media con estratos:", num(x_estratos_relev, 0))
print("Media simple de los 35 ingresos:", num(relevamiento["ingreso"].mean(), 0))
````

# 6. El intervalo de confianza para la media (recomendado)

*Unidad 6, secciones 5 y 6.*


Un intervalo de confianza se arma siempre con tres piezas:

```
estimación  ±  coeficiente × error standard
```

Con `σ` desconocido, el error standard es `s / √n` y el coeficiente sale de la t de Student con
`n − 1` grados de libertad. Al producto se lo llama **margen de error**, `e`.

### Cómo se hace

El ejemplo reproduce el intervalo de la sección 6.3 de la unidad 6, con los 35 ingresos del
relevamiento:

- `.dropna()` descarta las dos celdas vacías, los dos hogares que no declararon ingreso.
- `stats.t.ppf(0.975, gl)` es el **cálculo inverso** de la sección 6.7 de la unidad 3: el valor de la
  t que deja el 97,5% del área a su izquierda, y el 2,5% a su derecha. Con un 2,5% en cada cola
  queda el 95% en el medio.


````python
ingreso_relev = relevamiento["ingreso"].dropna()
n_relev = len(ingreso_relev)

ee_relev = ingreso_relev.std() / np.sqrt(n_relev)
t_relev = stats.t.ppf(0.975, n_relev - 1)
e_relev = t_relev * ee_relev

print("n =", n_relev, "  x̄ =", num(ingreso_relev.mean(), 0), "  s =", num(ingreso_relev.std(), 0))
print("Error standard:", num(ee_relev, 0), "  t:", num(t_relev, 4), "  margen:", num(e_relev, 0))
print(f"IC 95%: [{num(ingreso_relev.mean() - e_relev, 0)} ; {num(ingreso_relev.mean() + e_relev, 0)}]")
````

### Ahora vos

El intervalo del 95% para la media del ingreso, con la `muestra` de 200 ocupados: usá `x_barra`, `s`
y `n` del ejercicio 2. La línea que sigue al intervalo hace lo que en un proyecto real no se puede:
se fija si el intervalo cubre a `μ`. Las dos últimas arman el intervalo del 99%, con el coeficiente
que deja un 0,5% en cada cola.


````python
ee = COMPLETAR / np.sqrt(COMPLETAR)
t_critico = stats.t.ppf(COMPLETAR, n - 1)
e = t_critico * COMPLETAR

inferior = x_barra - e
superior = COMPLETAR

print("Error standard:", num(ee, 0), "  t:", num(t_critico, 4), "  margen:", num(e, 0))
print(f"IC 95%: [{num(inferior, 0)} ; {num(superior, 0)}]")
print(f"¿Cubre a μ = {num(mu, 0)}?", "sí" if inferior <= mu <= superior else "no")

e_99 = stats.t.ppf(0.995, n - 1) * ee
print(f"IC 99%: [{num(x_barra - e_99, 0)} ; {num(x_barra + e_99, 0)}]")
````

# 7. Qué significa el 95%

*Unidad 6, secciones 5.2 y 6.2.*


El 95% describe al **procedimiento**: si se repitiera el muestreo muchas veces, el 95% de los
intervalos cubriría al parámetro, siempre que se cumplan los supuestos del intervalo. El intervalo
de una muestra concreta lo cubre o no lo cubre. Con la población a mano, eso se puede contar.

### Cómo se hace

`intervalos_de_muestras(columna, n, repeticiones=100)` saca 100 muestras y arma el intervalo del 95%
de cada una, con la cuenta del ejercicio 6. Devuelve una tabla con las columnas `media`, `inferior`
y `superior`.

Para preguntar dos cosas a la vez sobre una columna se usa `&`, que quiere decir "y", con cada
condición entre paréntesis. `.sum()` cuenta los verdaderos y `.mean()` da la proporción.
`dibujar_intervalos(...)` pinta en verde azulado los intervalos que cubren y en coral los que
fallan.

Cien intervalos alcanzan para verlos, y no para medir la cobertura: de una tanda de cien a otra, la
cantidad que falla cambia bastante por azar. Por eso la segunda parte saca 5.000.


````python
mu_edad = poblacion["edad"].mean()
intervalos_edad = intervalos_de_muestras(poblacion["edad"], n=200, repeticiones=100)
cubre_edad = (intervalos_edad["inferior"] <= mu_edad) & (mu_edad <= intervalos_edad["superior"])
print("Intervalos que cubren a μ:", cubre_edad.sum(), "de", len(cubre_edad))
dibujar_intervalos(intervalos_edad, mu_edad, "Edad, 100 intervalos del 95%")

muchos_edad = intervalos_de_muestras(poblacion["edad"], n=200, repeticiones=5000)
cobertura_edad = ((muchos_edad["inferior"] <= mu_edad) & (mu_edad <= muchos_edad["superior"])).mean()
print("Cobertura en 5.000 intervalos:", num(100 * cobertura_edad, 1) + "%")
````

### Ahora vos

Lo mismo con el ingreso: primero 100 intervalos con `n = 200`, para verlos. Después 5.000, para
medir la cobertura. Y por último 5.000 con muestras de 30. Las dos últimas líneas cuentan, de los
5.000 intervalos con `n = 200`, cuántos fallan por quedar enteros debajo de `μ` y cuántos por quedar

# 8. La proporción: intervalo y test

*Unidad 6, sección 7; unidad 7, secciones 6 y 10.*


Para una proporción, el intervalo usa la normal y el error standard `√[p̂ (1 − p̂) / n]`. El test de
`H₀ : p = p₀` usa otro error standard, `√[p₀ (1 − p₀) / n]`, porque todo se calcula suponiendo que
`H₀` es cierta. La regla de decisión: si el valor p es menor o igual que α = 0,05, se rechaza
`H₀`. El ejercicio 11 explica el test paso a paso: si todavía no lo hiciste, conviene hacerlo antes
que este.

### Cómo se hace

El ejemplo reproduce las dos cuentas de los hogares sin gas, 20 de 37: el intervalo de la sección 7
de la unidad 6 y el test de la sección 6 de la unidad 7.

- `stats.norm.ppf(0.975)` es el cálculo inverso en la normal standard: da 1,96.
- `stats.norm.sf(z)` es el área a la **derecha** de `z`. El valor p de un test bilateral es el
  doble del área que queda más allá de `|z|`, y `abs(...)` saca el signo.
- `x ** 2` es `x` al cuadrado. `p * (1 - p)` usa `*` para multiplicar.


````python
p_relev = relevamiento["sin_gas"].mean()
z_95 = stats.norm.ppf(0.975)

ee_relev_p = np.sqrt(p_relev * (1 - p_relev) / 37)
print(f"p̂ = {num(p_relev, 4)}   IC 95%: [{num(p_relev - z_95 * ee_relev_p, 4)} ; "
      f"{num(p_relev + z_95 * ee_relev_p, 4)}]")

ee_relev_p0 = np.sqrt(0.5 * (1 - 0.5) / 37)
z_relev = (p_relev - 0.5) / ee_relev_p0
valor_p_relev_p = 2 * stats.norm.sf(abs(z_relev))
print("Test de H₀: p = 0,5   z =", num(z_relev, 4), "  valor p =", num(valor_p_relev_p, 4))
````

Con la `muestra`: el intervalo del 95% para la proporción de mujeres, y el test de H₀ : p = 0,5
(en el código, `0.5`), la afirmación de que entre los 16.709 ocupados de la población hay tantas
mujeres como varones. Las condiciones de la aproximación normal se cumplen de sobra con 200 casos:
`n p̂ ≥ 5` y `n (1 − p̂) ≥ 5` para el intervalo (sección 7 de la unidad 6), y `n p₀ ≥ 5` y
`n (1 − p₀) ≥ 5` para el test (sección 6 de la unidad 7).


````python
ee_p = np.sqrt(p_sombrero * (1 - COMPLETAR) / n)
inferior_p = p_sombrero - z_95 * ee_p
superior_p = p_sombrero + z_95 * COMPLETAR

p0 = COMPLETAR
ee_p0 = np.sqrt(p0 * (1 - p0) / COMPLETAR)
z_obs = (p_sombrero - COMPLETAR) / ee_p0
valor_p = 2 * stats.norm.sf(abs(z_obs))

print(f"p̂ = {num(p_sombrero, 4)}   IC 95%: [{num(inferior_p, 4)} ; {num(superior_p, 4)}]")
print("Test de H₀: p = 0,5   z =", num(z_obs, 4), "  valor p =", num(valor_p, 4))
print("La proporción verdadera es", num(p_mujeres, 4))
````

# 11. El test para la media y el valor p (recomendado)

*Unidad 7, secciones 2, 3, 4.1 y 5.*


Un test supone que `H₀` es cierta y calcula cuán raro sería el dato bajo ese supuesto. Para la
media, el estadístico es `t = (x̄ − μ₀) / (s / √n)`, con `n − 1` grados de libertad. Si el valor p
es menor o igual que `α`, se rechaza `H₀`; si no, no se rechaza, que **no es lo mismo que
aceptarla**.

### Cómo se hace

El ejemplo reproduce el test de la constructora de la sección 5 de la unidad 7, `H₀ : μ = 180.000`,
con los 35 ingresos del relevamiento y el error standard del ejemplo del ejercicio 6, que la celda
vuelve a calcular:

- `stats.t.sf(t, gl)` es el área a la derecha de `t` en la t de Student. El valor p bilateral es
  el doble del área más allá de `|t|`.
- `stats.ttest_1samp(datos, mu0)` hace el test entero de una vez. Devuelve el estadístico en
  `.statistic` y el valor p en `.pvalue`: sirve para controlar la cuenta a mano.


````python
ingreso_relev = relevamiento["ingreso"].dropna()
n_relev = len(ingreso_relev)
ee_relev = ingreso_relev.std() / np.sqrt(n_relev)

mu0_relev = 180000
t_obs_relev = (ingreso_relev.mean() - mu0_relev) / ee_relev
valor_p_relev = 2 * stats.t.sf(abs(t_obs_relev), n_relev - 1)

print("t =", num(t_obs_relev, 4), "  valor p =", num(valor_p_relev, 4))
print("¿Se rechaza H₀ con α = 0,05?", "sí" if valor_p_relev <= 0.05 else "no")

prueba_relev = stats.ttest_1samp(ingreso_relev, mu0_relev)
print("Con scipy: t =", num(prueba_relev.statistic, 4), "  valor p =", num(prueba_relev.pvalue, 4))
````

# 12. Los dos errores y la potencia (recomendado)

*Unidad 7, secciones 3.2, 4 y 5.*


El **error de tipo I** es rechazar una `H₀` cierta, y su probabilidad es `α` si se cumplen los
supuestos del test. El **error de tipo II** es no rechazar una `H₀` falsa. La **potencia** es la
probabilidad de rechazar cuando `H₀` es falsa: la capacidad de detectar una diferencia que existe.
La proporción de errores de tipo II es `1 − potencia`. Con una sola muestra no se ven; con mil, sí.

### Cómo se hace

El ejemplo usa la edad. Con `mu0 = mu_edad`, `H₀` es cierta, y cada rechazo es un error de tipo I.
Con `mu0 = mu_edad + 2`, `H₀` es falsa, y la proporción de rechazos es la potencia.


````python
mu_edad = poblacion["edad"].mean()
tipo_1_edad = proporcion_de_rechazos(poblacion["edad"], mu0=mu_edad, n=50)
potencia_edad = proporcion_de_rechazos(poblacion["edad"], mu0=mu_edad + 2, n=50)

print("H₀ cierta, μ₀ = μ:           rechaza", num(100 * tipo_1_edad, 1) + "%")
print("H₀ falsa, μ₀ = μ + 2 años:   rechaza", num(100 * potencia_edad, 1) + "%")
````

# 13. Dos grupos: la diferencia de medias

*Unidad 7, secciones 9 y 10.*


Para comparar las medias de dos grupos independientes, la sección 9.5 de la unidad 7 recomienda el

para la diferencia, que dice de qué tamaño es.

### Cómo se hace

El ejemplo compara Ramallo con Córdoba, como las secciones 9.3 a 9.5 de la unidad 7:

- `stats.ttest_ind(grupo_1, grupo_2, equal_var=False)` es el test de Welch. Además de
  `.statistic` y `.pvalue` devuelve `.df`, los grados de libertad de Welch.
- `.var()` es la varianza muestral, con divisor `n − 1`. El error standard de la diferencia es
  `√(s₁²/n₁ + s₂²/n₂)`: las varianzas se suman, como en la sección 5.6 de la unidad 2.
- El intervalo usa ese error standard de Welch. Por eso no da el de la sección 9.4 de la unidad 7,
  que usa el desvío combinado.


````python
ramallo_relev = relevamiento[relevamiento["localidad"] == "Ramallo"]["ingreso"].dropna()
cordoba_relev = relevamiento[relevamiento["localidad"] == "Córdoba"]["ingreso"].dropna()

welch_relev = stats.ttest_ind(ramallo_relev, cordoba_relev, equal_var=False)
print("Welch: t =", num(welch_relev.statistic, 4), "  gl =", num(welch_relev.df, 2),
      "  valor p =", num(welch_relev.pvalue, 4))

dif_relev = ramallo_relev.mean() - cordoba_relev.mean()
ee_dif_relev = np.sqrt(ramallo_relev.var() / len(ramallo_relev) + cordoba_relev.var() / len(cordoba_relev))
e_dif_relev = stats.t.ppf(0.975, welch_relev.df) * ee_dif_relev
print(f"Diferencia: {num(dif_relev, 0)}   IC 95%: [{num(dif_relev - e_dif_relev, 0)} ; "
      f"{num(dif_relev + e_dif_relev, 0)}]")
````

````python
pares = panel.sample(n=COMPLETAR, random_state=8102)

apareado = stats.ttest_rel(pares["ingreso_2t"], pares["COMPLETAR"])
independiente = stats.ttest_ind(pares["ingreso_2t"], pares["ingreso_1t"], equal_var=False)
dif_media_pares = (pares["ingreso_2t"] - pares["ingreso_1t"]).COMPLETAR()

print("Aumento medio:", num(dif_media_pares, 0), "pesos, un",
      num(100 * dif_media_pares / pares["ingreso_1t"].mean(), 1) + "% del ingreso medio del primer trimestre")
print("Apareado:      t =", num(apareado.statistic, 4), "  valor p =", num(apareado.pvalue, 4))
print("Independiente: t =", num(independiente.statistic, 4), "  valor p =", num(independiente.pvalue, 4))
print("Desvío del ingreso: primer trimestre", num(pares["ingreso_1t"].std(), 0),
      "  segundo trimestre", num(pares["ingreso_2t"].std(), 0),
      "  de los cambios", num((pares["ingreso_2t"] - pares["ingreso_1t"]).std(), 0))
````

# 15. La dispersión: la varianza y el cociente

*Unidad 6, sección 8; unidad 7, secciones 7 y 8.*


El intervalo para la varianza usa la chi cuadrado, y exige que la población sea normal. La sección 8
de la unidad 6 advierte que ese supuesto **no se relaja con `n` grande**. Con la población a mano se
puede ver qué pasa cuando no se cumple.

### Cómo se hace

El ejemplo reproduce el intervalo para `σ` de la sección 8 de la unidad 6 y el test de
`H₀ : σ = 60.000` de la sección 7 de la unidad 7:

- `stats.chi2.ppf(0.975, gl)` y `stats.chi2.ppf(0.025, gl)` son los dos cuantiles de la chi
  cuadrado. Los denominadores van cruzados: el cuantil grande da el extremo chico.
- El intervalo para `σ` es la raíz del intervalo para `σ²`. `gl_relev` son los grados de libertad,
  `n − 1`, con los 35 ingresos del relevamiento.
- `stats.chi2.cdf(...)` es el área a la izquierda y `stats.chi2.sf(...)` el área a la derecha. Como
  la chi cuadrado no es simétrica, el valor p es el doble de la más chica de las dos, y `min(...)`
  elige esa.


````python
ingreso_relev = relevamiento["ingreso"].dropna()
gl_relev = len(ingreso_relev) - 1
inferior_relev = np.sqrt(gl_relev * ingreso_relev.var() / stats.chi2.ppf(0.975, gl_relev))
superior_relev = np.sqrt(gl_relev * ingreso_relev.var() / stats.chi2.ppf(0.025, gl_relev))
print(f"IC 95% para σ: [{num(inferior_relev, 0)} ; {num(superior_relev, 0)}]")

chi_relev = gl_relev * ingreso_relev.var() / 60000 ** 2
valor_p_chi = 2 * min(stats.chi2.cdf(chi_relev, gl_relev), stats.chi2.sf(chi_relev, gl_relev))
print("Test de H₀: σ = 60.000   χ² =", num(chi_relev, 4), "  valor p =", num(valor_p_chi, 4))
````
